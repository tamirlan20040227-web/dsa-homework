````python
import csv
with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))
print(len(rows), "строк")
````

## [База]
1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).
2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
3. .Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой
4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

````python
# База
````

````python
#1
def to_num(s):
    s = s.replace(" ", "")
    return float(s)
print(to_num(rows[0]["amount"]))
def clean_city(s):
    s = s.strip().lower()

    city_names = {
        "almaty": "Алматы",
        "astana": "Астана",
        "shymkent": "Шымкент"
    }

    return city_names.get(s, s.title())

cities_clean = set()

for row in rows:
    city = clean_city(row["city"])
    cities_clean.add(city)

print(cities_clean)
print("Количество городов:", len(cities_clean))
````

````python
#2
category_counts = {}
for row in rows:
    category = row ["category"]

    if category not in category_counts:
        category_counts[category] = 0

    category_counts[category] += 1
sorted_categories = sorted(
    category_counts.items(),
    key = lambda x: x[1],
    reverse = True
)
print(category_counts)
print(sorted_categories)
````

````python
category_revenue = {}
category_count = {}

for row in rows:
    if row["status"] == "оплачен":
        category = row["category"]
        amount = to_num(row["amount"])

        if category not in category_revenue:
            category_revenue[category] = 0
            category_count[category] = 0

        category_revenue[category] += amount
        category_count[category] += 1

print(f"{'Категория':<15} | {'Выручка':>15} | {'Средний чек':>15}")
print("-" * 51)

for category in category_revenue:
    revenue = category_revenue[category]
    average = revenue / category_count[category]

    print(f"{category:<15} | {revenue:>15.0f} | {average:>15.0f}")
````

````python
#4
month_counts = {}

for row in rows:
    month = row["order_date"][:7]

    if month not in month_counts:
        month_counts[month] = 0

    month_counts[month] += 1

for month in sorted(month_counts):
    print(f"{month}: {month_counts[month]} заказов")
````

## [Стандарт]
5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.
7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.

````python
# Стандарт
````

````python
month_revenue = {}

for row in rows:
    if row["status"] == "оплачен":
        month = row["order_date"][:7]
        amount = to_num(row["amount"])

        if month not in month_revenue:
            month_revenue[month] = 0

        month_revenue[month] += amount

september = month_revenue["2026-09"]
october = month_revenue["2026-10"]

change = (october - september) / september * 100

print(f"Сентябрь: {september:.0f} ₸")
print(f"Октябрь: {october:.0f} ₸")
print(f"Изменение: {change:.1f}%")
````

````python
city_orders = {}
city_cancelled = {}

for row in rows:
    city = clean_city(row["city"])

    if city not in city_orders:
        city_orders[city] = 0
        city_cancelled[city] = 0

    city_orders[city] += 1

    if row["status"] == "отменён":
        city_cancelled[city] += 1

for city in sorted(city_orders):
    percent = city_cancelled[city] / city_orders[city] * 100
    print(f"{city:<12} | {percent:.1f}%")
````

````python
customer_revenue = {}

for row in rows:
    if row["status"] == "оплачен":
        customer = row["customer_id"]
        amount = to_num(row["amount"])

        if customer not in customer_revenue:
            customer_revenue[customer] = 0

        customer_revenue[customer] += amount

top_customers = sorted(
    customer_revenue.items(),
    key=lambda x: x[1],
    reverse=True
)

print("ТОП-5 клиентов:")

for customer, revenue in top_customers[:5]:
    print(f"{customer:<10} | {revenue:.0f} ₸")
````

````python
city_customers = {}

for row in rows:
    city = clean_city(row["city"])
    customer = row["customer_id"]

    if city not in city_customers:
        city_customers[city] = set()

    city_customers[city].add(customer)

for city in sorted(city_customers):
    print(f"{city:<12} | {len(city_customers[city])}")
````
